## 1. Exploratory Data Analysis & Business Insights

**Category Performance Analysis**


```python
category_performance = df.groupby("Category").agg({
    "Sales": "sum",
    "Profit": "sum"
})
```

```python
category_performance
```

**Business Insight:**

Technology generates the highest total sales and total profit among the three product categories.

#### Category Sales and Profit — Visualisation

# Visualise total sales and profit by product category

```python
category_performance.plot(
    kind="bar",
    figsize=(8, 5),
    title="Sales and Profit by Category"
)

plt.xlabel("Category")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

**Business Question:** Which product category generates the highest total sales and profit?

**Method:** Grouped the data by Category and calculated the sum of Sales and Profit using groupby() and agg().

**Result:** Technology generates the highest total sales (839,893.28) and profit (146,543.38) among the three categories.

**Interpretation:** Technology is the strongest contributor to overall category sales and profit.

**Caveat:** This chart shows category-level totals. It does not tell us whether every individual product is profitable.

# Analyse sales and profit for each Furniture sub-category

```python
furniture_df = df[df["Category"] == "Furniture"]

sub_category_performance = furniture_df.groupby("Sub-Category").agg({
    "Sales": "sum",
    "Profit": "sum"
})
```

```python
sub_category_performance
```

```python
furniture_df = df[df["Category"] == "Furniture"]

sub_category_performance = furniture_df.groupby("Sub-Category").agg({
    "Sales": "mean",
    "Discount": "mean",
    "Profit": "mean"
})
```

```python
sub_category_performance
```

**Business Insight:**

Tables have the highest average discount (25.81%) and the lowest average profit (-54.46) among Furniture subcategories. Bookcases also generate negative average profit. Chairs are the most profitable Furniture subcategory by average profit.

# Visualise average profit by Furniture subcategory

```python
plt.figure(figsize=(8,5))
plt.bar(
    sub_category_performance.index,
    sub_category_performance["Profit"]
)
plt.title("Average Profit by Furniture Subcategory")
plt.xlabel("Subcategory")
plt.ylabel("Average Profit")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

**Business Question:** How does each Furniture subcategory perform in terms of average profit?

**Method:** Filtered the dataset to include only Furniture records, grouped the data by Sub-Category, and calculated average Sales, Discount, and Profit.

**Result:** Tables have the lowest average profit (-54.46), followed by Bookcases (-15.66). Chairs have the highest average profit (42.94).

**Interpretation:** Tables and Bookcases generate negative average profit, while Chairs and Furnishings generate positive average profit. Tables also have the highest average discount (25.81%).

**Caveat:** Average profit and discount values show an association, not causation. Higher discounts cannot be assumed to be the reason for lower profit.

### 4. Monthly Sales Trend Analysis

Analyse monthly sales to identify trends and unusual changes over time.

# Calculate total sales by month

```python
monthly_sales = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Sales"].sum()

monthly_sales
```

#### Monthly Sales Trend — Visualisation

```python
plt.figure(figsize=(14, 5))

plt.plot(monthly_sales.index.astype(str), monthly_sales.values)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Business Insight:**

Monthly sales fluctuate significantly throughout the four-year period. November 2026 records the highest monthly sales, while February 2023 records the lowest. Sales appear to increase overall in the later period, but further investigation is needed to understand the reasons behind these fluctuations.

## 6. Identify Highest and Lowest Sales Months

# Identify the months with the highest and lowest sales

```python
highest_sales_months = monthly_sales.sort_values(ascending=False).head(5)
lowest_sales_months = monthly_sales.sort_values().head(5)

print("Top 5 Months:")
display(highest_sales_months)

print("Bottom 5 Months:")
display(lowest_sales_months)
```

**Business Insight:**

Monthly sales show considerable variation. November 2026 recorded the highest sales (118,454.51), while February 2023 recorded the lowest (4,519.89). January and February frequently appear among the lowest-sales months, whereas September to December appear among the highest. This may suggest seasonal variation, but further investigation is needed to confirm the pattern.

# Visualise the top 5 and bottom 5 months by sales

```python
monthly_extremes = pd.concat([
    lowest_sales_months,
    highest_sales_months
])

plt.figure(figsize=(12, 5))
plt.bar(monthly_extremes.index.astype(str), monthly_extremes.values)

plt.title("Top 5 and Bottom 5 Months by Sales")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

**Business Question:** Which months recorded the highest and lowest sales?

**Method:** Grouped sales by month and selected the top five and bottom five months.

**Result:** November 2026 had the highest sales (118,454.51), while February 2023 had the lowest (4,519.89).

**Interpretation:** Sales vary considerably across months. January and February appear frequently among the lowest-sales months.

**Caveat:** Monthly totals can be affected by the number of orders and other factors. The pattern does not establish seasonality or its cause.
